# Data Understanding

```python
from pathlib import Path

import pandas as pd

WORKBOOK = Path("..") / "data" / "raw" / "kenya_maize_production_2019_2025_v2.xlsx"
```

## 1. Workbook overview

```python
# how many sheets are in the workbook and their names
xls = pd.ExcelFile(WORKBOOK)
print(xls.sheet_names)
```

```python
for sheet in xls.sheet_names:
    print("=" * 60)
    df = pd.read_excel(xls, sheet_name=sheet, header=None)
    print(f"Sheet '{sheet}' loaded with shape: {df.shape}")
    print(df.head(10).to_string())
```

## 2. The READ ME sheet

```python
readme = pd.read_excel(xls, sheet_name="READ ME", header=None)
print(readme.to_string())
```

## 3. Loading the data sheets

```python
national_es = pd.read_excel(xls, sheet_name="National_ES_2021_2025")
supply_use = pd.read_excel(xls, sheet_name="Supply_use_2024_2025")
national_afa = pd.read_excel(xls, sheet_name="National_AFA_2019_2024")
county = pd.read_excel(xls, sheet_name="County_2020_2024")
```

```python
for name, df in {"national_es": national_es, "supply_use": supply_use,
                 "national_afa": national_afa, "county": county}.items():
    print("=" * 60)
    print(name)
    df.info()
```
